```python
# 导入必要的库
import pandas as pd
from nba_api.stats.endpoints import playercareerstats

# LeBron James 的 NBA 官方球员 ID
lebron_id = '2544'

# 获取 LeBron James 的职业生涯统计数据
lebron = playercareerstats.PlayerCareerStats(player_id=lebron_id)
lebron_df = lebron.get_data_frames()[0]

# 展示职业生涯统计数据
print(lebron_df)

# 将数据保存到 CSV 文件
lebron_df.to_csv('lebron_career_stats.csv', index=False)

# 选取部分列进行查看
columns_to_display = ['SEASON_ID', 'TEAM_ID', 'TEAM_ABBREVIATION', 'GP', 'PTS', 'AST', 'REB']
lebron_summary = lebron_df[columns_to_display]

# 打印精简版数据
print(lebron_summary)

# 统计 LeBron 的生涯场均得分、助攻、篮板
lebron_avg_stats = lebron_summary[['PTS', 'AST', 'REB']].mean()
print("\nLeBron James Career Averages:")
print(lebron_avg_stats)
```

```python
from nba_api.stats.endpoints import SynergyPlayTypes
import pandas as pd

# LeBron James的球员ID
lebron_id = '2544'  # LeBron James的NBA ID

# 获取LeBron的所有进攻和防守play type数据
def get_lebron_playtypes(season='2023-24', per_mode='PerGame', season_type='Regular Season'):
    play_types = [
        'Isolation', 'Transition', 'PRBallHandler', 'PRRollman', 'Postup', 
        'Spotup', 'Handoff', 'Cut', 'OffScreen', 'OffRebound', 'Misc'
    ]
    
    all_data = []
    
    for play_type in play_types:
        try:
            synergy_data = SynergyPlayTypes(
                league_id='00',
                per_mode_simple=per_mode,
                play_type_nullable=play_type,
                player_or_team_abbreviation='P',
                season_type_all_star=season_type,
                season=season,
                type_grouping_nullable='Offensive'  # 可以更改为Defensive来获取防守数据
            )
            
            df = synergy_data.get_data_frames()[0]
            df = df[df['PLAYER_ID'].astype(str) == lebron_id]  # 过滤LeBron数据
            df['PLAY_TYPE'] = play_type  # 添加战术类型
            all_data.append(df)
            print(f"Fetched data for play type: {play_type}")

        except Exception as e:
            print(f"Error fetching data for {play_type}: {e}")

    # 合并所有play type的数据
    if all_data:
        result_df = pd.concat(all_data, ignore_index=True)
        result_df.to_csv('lebron_synergy_playtypes.csv', index=False)
        print("Data saved to lebron_synergy_playtypes.csv")
        return result_df
    else:
        print("No data fetched.")
        return None

# 执行数据抓取
lebron_playtype_data = get_lebron_playtypes()

# 打印部分数据
if lebron_playtype_data is not None:
    print(lebron_playtype_data.head())
```
