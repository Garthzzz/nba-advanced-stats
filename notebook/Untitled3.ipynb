```python
import pandas as pd
from nba_api.stats.endpoints import leaguegamefinder, playbyplay, playbyplayv2, playbyplayv3
import time

# Step 1: 获取孟菲斯灰熊最近一场比赛
def get_latest_grizzlies_game():
    print("Fetching latest Grizzlies game...")
    gamefinder = leaguegamefinder.LeagueGameFinder(team_id_nullable='1610612763', season_nullable='2023-24')
    games = gamefinder.get_data_frames()[0]

    # 按照日期排序，获取最近的一场比赛
    latest_game = games.sort_values('GAME_DATE', ascending=False).iloc[0]
    game_id = latest_game['GAME_ID']
    print(f"Latest game ID: {game_id}")
    return game_id

# Step 2: 获取比赛日志数据
def fetch_play_by_play_data(game_id):
    # 获取 PlayByPlay v1 数据
    print("Fetching PlayByPlay V1 data...")
    pbp_v1 = playbyplay.PlayByPlay(game_id=game_id).get_data_frames()[0]
    pbp_v1.to_csv('v1.csv', index=False)
    print("PlayByPlay V1 data saved to v1.csv")

    time.sleep(2)  # 避免过快请求，防止限流

    # 获取 PlayByPlay V2 数据
    print("Fetching PlayByPlay V2 data...")
    pbp_v2 = playbyplayv2.PlayByPlayV2(game_id=game_id).get_data_frames()[0]
    pbp_v2.to_csv('v2.csv', index=False)
    print("PlayByPlay V2 data saved to v2.csv")

    time.sleep(2)

    # 获取 PlayByPlay V3 数据
    print("Fetching PlayByPlay V3 data...")
    pbp_v3 = playbyplayv3.PlayByPlayV3(game_id=game_id).get_data_frames()[0]
    pbp_v3.to_csv('v3.csv', index=False)
    print("PlayByPlay V3 data saved to v3.csv")

    return pbp_v1, pbp_v2, pbp_v3

# 主程序执行
if __name__ == "__main__":
    latest_game_id = get_latest_grizzlies_game()
    pbp_v1, pbp_v2, pbp_v3 = fetch_play_by_play_data(latest_game_id)

    # 打印数据框的前几行预览
    print("\nV1 PlayByPlay Data Preview:")
    print(pbp_v1.head())

    print("\nV2 PlayByPlay Data Preview:")
    print(pbp_v2.head())

    print("\nV3 PlayByPlay Data Preview:")
    print(pbp_v3.head())
```
