```python
import pandas as pd
import os
import time
from nba_api.stats.endpoints import synergyplaytypes

# 定义读取的参数
SEASON = "2022-23"
PER_MODE = "Totals"
SEASON_TYPE = "Regular Season"
LEAGUE_ID = "00"

# 战术类型
PLAY_TYPES = [
    'Isolation', 'Transition', 'PRBallHandler', 'PRRollman', 'Postup',
    'Spotup', 'Handoff', 'Cut', 'OffScreen', 'OffRebound', 'Misc'
]

# 战术分组类型
TYPE_GROUPINGS = ["Offensive", "Defensive"]

# 函数：获取 Synergy Play Type 数据
def fetch_and_save_synergy_data(play_type, type_grouping):
    print(f"Fetching data for Play Type: {play_type}, Type Grouping: {type_grouping}")

    try:
        synergy_data = synergyplaytypes.SynergyPlayTypes(
            league_id=LEAGUE_ID,
            per_mode_simple=PER_MODE,
            play_type_nullable=play_type,
            player_or_team_abbreviation="P",
            season_type_all_star=SEASON_TYPE,
            season=SEASON,
            type_grouping_nullable=type_grouping
        )

        df_player_stats = synergy_data.get_data_frames()[0]  # 获取球员数据

        if df_player_stats.empty:
            print(f"Warning: No data found for Play Type {play_type} ({type_grouping})")
            return

        # 构造文件名
        file_name = f"synergy_{type_grouping.lower()}_{play_type.lower()}_player_stats_2022-23.csv"
        
        # 保存数据到本地
        df_player_stats.to_csv(file_name, index=False)
        print(f"Successfully saved player data to {file_name}")

    except Exception as e:
        print(f"Error fetching data for Play Type {play_type} ({type_grouping}): {e}")

# 主程序：遍历所有战术类别和进攻/防守类型
def main():
    for type_grouping in TYPE_GROUPINGS:
        for play_type in PLAY_TYPES:
            fetch_and_save_synergy_data(play_type, type_grouping)
            time.sleep(1)  # 延迟以避免请求频繁

    print("Synergy Play Type data fetching completed.")

# 运行主程序
if __name__ == "__main__":
    main()
```
