```python
#  BoxScoreTraditionalV2， PlayerStats、TeamStarterBenchStats 和 TeamStats 

import os
import pandas as pd
from glob import glob
from nba_api.stats.endpoints import LeagueGameFinder, BoxScoreTraditionalV2

# ===================== 常量定义 =====================

SEASON = "2022-23"

# 输出目录
OUTPUT_DIR_PLAYER = "boxscore_traditional_player_stats"
OUTPUT_DIR_STARTER_BENCH = "boxscore_traditional_team_starter_bench_stats"
OUTPUT_DIR_TEAM = "boxscore_traditional_team_stats"

# 已处理比赛 ID 文件
PROCESSED_FILE_PLAYER = "processed_games_player_stats.txt"
PROCESSED_FILE_STARTER_BENCH = "processed_games_team_starter_bench_stats.txt"
PROCESSED_FILE_TEAM = "processed_games_team_stats.txt"

# ===================== 函数定义 =====================

def get_season_game_ids(season=SEASON):
    """
    获取指定赛季的所有常规赛比赛 ID 列表及每场比赛的比赛日期
    """
    print("Fetching all games...")
    finder = LeagueGameFinder(season_nullable=season, season_type_nullable='Regular Season')
    games_df = finder.get_data_frames()[0]
    
    print("Columns in the games data:", games_df.columns)
    # 过滤 2022-23 赛季（SEASON_ID 以 '2022' 结尾）
    games_df = games_df[games_df['SEASON_ID'].astype(str).str.endswith('2022')]
    # 去重
    games_df = games_df.drop_duplicates(subset=['GAME_ID'])
    # 保留 NBA 官方队伍
    valid_teams = [
        'ATL','BOS','BKN','CHA','CHI','CLE','DAL','DEN','DET',
        'GSW','HOU','IND','LAC','LAL','MEM','MIA','MIL','MIN',
        'NOP','NYK','OKC','ORL','PHI','PHX','POR','SAC','SAS',
        'TOR','UTA','WAS'
    ]
    games_df = games_df[
        games_df['MATCHUP'].apply(
            lambda x: x.split(' ')[0] in valid_teams and x.split(' ')[-1] in valid_teams
        )
    ]
    print(f"Filtered regular season games count: {len(games_df)}")
    
    game_ids = games_df['GAME_ID'].tolist()
    # 建立 GAME_ID -> GAME_DATE 映射
    game_date_map = dict(zip(games_df['GAME_ID'], games_df['GAME_DATE']))
    print(f"Total games found for {season}: {len(game_ids)}")
    return game_ids, game_date_map

def load_processed_games(file_path):
    """
    读取已处理比赛 ID 列表
    """
    if os.path.exists(file_path):
        with open(file_path, "r") as f:
            return set(line.strip() for line in f)
    return set()

def save_processed_game(file_path, game_id):
    """
    追加保存已处理的比赛 ID
    """
    with open(file_path, "a") as f:
        f.write(f"{game_id}\n")

def save_data(df, output_dir, game_date, prefix):
    """
    按月份保存 DataFrame 到 CSV
    """
    # 按月份组织文件
    month = pd.to_datetime(game_date).strftime('%Y-%m')
    file_name = f"{prefix}_{SEASON}_{month}.csv"
    file_path = os.path.join(output_dir, file_name)
    
    if os.path.exists(file_path):
        df.to_csv(file_path, mode='a', header=False, index=False)
    else:
        df.to_csv(file_path, index=False)

def remove_duplicate_rows(file_path):
    """
    去除 CSV 文件中的重复行，并打印去重前后行数
    """
    try:
        df = pd.read_csv(file_path)
        original = len(df)
        df_clean = df.drop_duplicates(keep='first')
        df_clean.to_csv(file_path, index=False)
        print(f"Successfully removed duplicates and saved: {file_path}")
        print(f"Original rows: {original}, After deduplication: {len(df_clean)}")
    except Exception as e:
        print(f"Error processing {file_path}: {e}")

def process_games(game_ids, game_date_map):
    """
    对每场比赛依次调用 BoxScoreTraditionalV2，保存三个数据集
    """
    # 读取已处理记录
    processed_p = load_processed_games(PROCESSED_FILE_PLAYER)
    processed_sb = load_processed_games(PROCESSED_FILE_STARTER_BENCH)
    processed_t = load_processed_games(PROCESSED_FILE_TEAM)
    
    # 确保输出目录存在
    for d in [OUTPUT_DIR_PLAYER, OUTPUT_DIR_STARTER_BENCH, OUTPUT_DIR_TEAM]:
        if not os.path.exists(d):
            os.makedirs(d)
    
    for game_id in game_ids:
        # 如果此场比赛三个数据集都已处理，跳过
        if game_id in processed_p and game_id in processed_sb and game_id in processed_t:
            continue
        
        print(f"Processing GameID {game_id} ...")
        game_date = game_date_map.get(game_id)
        
        try:
            box = BoxScoreTraditionalV2(game_id=game_id)
            dfs = box.get_data_frames()  # [PlayerStats, TeamStarterBenchStats, TeamStats]
        except Exception as e:
            print(f"Error fetching data for {game_id}: {e}")
            continue
        
        # 保存 PlayerStats
        if game_id not in processed_p:
            try:
                save_data(dfs[0], OUTPUT_DIR_PLAYER, game_date, "player_stats")
                save_processed_game(PROCESSED_FILE_PLAYER, game_id)
            except Exception as e:
                print(f"Error saving PlayerStats for {game_id}: {e}")
        
        # 保存 TeamStarterBenchStats
        if game_id not in processed_sb:
            try:
                save_data(dfs[1], OUTPUT_DIR_STARTER_BENCH, game_date, "starter_bench_stats")
                save_processed_game(PROCESSED_FILE_STARTER_BENCH, game_id)
            except Exception as e:
                print(f"Error saving TeamStarterBenchStats for {game_id}: {e}")
        
        # 保存 TeamStats
        if game_id not in processed_t:
            try:
                save_data(dfs[2], OUTPUT_DIR_TEAM, game_date, "team_stats")
                save_processed_game(PROCESSED_FILE_TEAM, game_id)
            except Exception as e:
                print(f"Error saving TeamStats for {game_id}: {e}")

# ===================== 主流程 =====================

if __name__ == "__main__":
    # 1. 获取赛季比赛列表及日期
    game_ids, game_date_map = get_season_game_ids()
    # 2. 抓取并保存数据
    process_games(game_ids, game_date_map)
    # 3. 去重所有输出目录下的 CSV 文件
    for out_dir in [OUTPUT_DIR_PLAYER, OUTPUT_DIR_STARTER_BENCH, OUTPUT_DIR_TEAM]:
        for csv_file in glob(os.path.join(out_dir, "*.csv")):
            remove_duplicate_rows(csv_file)


```
