```python
import pandas as pd
from nba_api.stats.endpoints import teamplayeronoffsummary

# 设置API参数
season = "2022-23"  # 赛季
season_type = "Regular Season"  # 只考虑常规赛
per_mode = "PerGame"  # 每场数据
measure_type = "Base"  # 基础统计
team_id = 0  # 0表示所有球队
pace_adjust = "N"  # 不调整节奏
plus_minus = "N"  # 不包含正负值
rank = "N"  # 不排序

# 抓取数据
team_on_off_data = teamplayeronoffsummary.TeamPlayerOnOffSummary(
    season=season,
    season_type_all_star=season_type,
    per_mode_detailed=per_mode,
    measure_type_detailed_defense=measure_type,
    team_id=team_id,
    pace_adjust=pace_adjust,
    plus_minus=plus_minus,
    rank=rank
)

# 获取数据并转换为DataFrame
df_overall = team_on_off_data.get_data_frames()[0]  # Overall Team Player On/Off Summary
df_players_on = team_on_off_data.get_data_frames()[1]  # Players On Court Summary
df_players_off = team_on_off_data.get_data_frames()[2]  # Players Off Court Summary

# 保存数据到CSV文件
df_overall.to_csv("teamplayeronoffsummary_overall_2022_23.csv", index=False)
df_players_on.to_csv("teamplayeronoffsummary_on_court_2022_23.csv", index=False)
df_players_off.to_csv("teamplayeronoffsummary_off_court_2022_23.csv", index=False)

# 显示前几行数据
print("Overall Team Player On/Off Summary:")
print(df_overall.head())

print("\nPlayers On Court Summary:")
print(df_players_on.head())

print("\nPlayers Off Court Summary:")
print(df_players_off.head())
```

```python
import pandas as pd
from nba_api.stats.endpoints import teamplayeronoffsummary
from nba_api.stats.static import teams

# 定义赛季和赛季类型
season = "2022-23"
season_type = "Regular Season"

# 获取所有 NBA 球队
nba_teams = teams.get_teams()
team_ids = [team['id'] for team in nba_teams]

# 创建空的 DataFrame 存储数据
df_players_on_all = pd.DataFrame()
df_players_off_all = pd.DataFrame()

# 遍历所有球队
for team_id in team_ids:
    print(f"Fetching data for team ID: {team_id}")
    
    # 获取当前球队的数据
    try:
        team_on_off_data = teamplayeronoffsummary.TeamPlayerOnOffSummary(
            team_id=team_id, season=season, season_type_all_star=season_type
        )
        
        # 提取 on/off 数据
        df_players_on = team_on_off_data.get_data_frames()[1]  # On court data
        df_players_off = team_on_off_data.get_data_frames()[2]  # Off court data
        
        # 累加数据
        df_players_on_all = pd.concat([df_players_on_all, df_players_on], ignore_index=True)
        df_players_off_all = pd.concat([df_players_off_all, df_players_off], ignore_index=True)
    
    except Exception as e:
        print(f"Error fetching data for team ID {team_id}: {e}")

# 将数据保存到 CSV 文件
df_players_on_all.to_csv("teamplayeronoffsummary_on_court_2022_23.csv", index=False)
df_players_off_all.to_csv("teamplayeronoffsummary_off_court_2022_23.csv", index=False)

# 打印前五行数据
print("Players On Court Data (Top 5 Rows):")
print(df_players_on_all.head())

print("\nPlayers Off Court Data (Top 5 Rows):")
print(df_players_off_all.head())
```
